```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(df["Final_CGPA"], bins=20, kde=True)

plt.title("Distribution of Final CGPA")
plt.xlabel("Final CGPA")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 3))

sns.boxplot(x=df["Final_CGPA"])

plt.title("Box Plot of Final CGPA")
plt.xlabel("Final CGPA")

plt.show()
```

```python
Q1 = df["Final_CGPA"].quantile(0.25)
Q3 = df["Final_CGPA"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Final_CGPA"] < lower_bound) |
    (df["Final_CGPA"] > upper_bound)
]

print("Number of potential outliers:", len(outliers))
```

```python
# Summarize low-end outliers without displaying individual student records or IDs.
outlier_summary = pd.DataFrame({
    "count": [len(outliers)],
    "minimum_Final_CGPA": [outliers["Final_CGPA"].min()],
    "maximum_Final_CGPA": [outliers["Final_CGPA"].max()],
})
outlier_summary.round(3)
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance_Pct",
    y="Final_CGPA"
)

plt.title("Attendance vs Final CGPA")
plt.xlabel("Attendance (%)")
plt.ylabel("Final CGPA")

plt.show()
```

```python
correlation = df["Attendance_Pct"].corr(df["Final_CGPA"])

print("Correlation between Attendance and Final CGPA:", correlation)
```

```python
numeric_cols = [
    "Age",
    "Attendance_Pct",
    "Study_Hours_Per_Day",
    "Previous_CGPA",
    "Sleep_Hours",
    "Social_Hours_Week",
    "Final_CGPA"
]

correlation_matrix = df[numeric_cols].corr()

correlation_matrix["Final_CGPA"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Features")

plt.show()
```
